## Task 2 - Vectorized Arithmetic and Linear Algebra
● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().
● For a square matrix, find its transpose and determinant.
● If invertible, find the inverse and verify A @ A^-1 using np.allclose().
● Use vectorized NumPy operations; avoid explicit Python loops.

```python
#Create 2 matrix
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
print("A:")
print(A)
print("\nB:")
print(B)
add = A+B
print("Addition of matrix: ", add)
mult= A*B
print("Multiplication of matrix elements: ", mult) #Element wise
mat_mult= A@B
print("Multiplication of matrix: ", mat_mult) #Matrix multiplication
transpose = A.T #Transpose
print("Transpose:")
print(transpose)
#Determinant
det = np.linalg.det(A)
print("Determinant:", det)
#Inverse
if np.linalg.det(A) != 0:

    inverse = np.linalg.inv(A)

    print("Inverse:")
    print(inverse)

    result = A @ inverse

    print("\nA @ A^-1:")
    print(result)

    print("\nVerification:")
    print(np.allclose(result, np.eye(A.shape[0])))

else:
    print("Matrix is not invertible.")
```

## Task 3 - Normal Distribution and Histogram
● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard deviation.
● Find the sample mean and sample standard deviation. Compare them with the chosen values.
● Plot a histogram with it.

```python
import matplotlib.pyplot as plt
np.random.seed(42)

mean = 50
std = 10

data = np.random.normal(mean, std, 1000)

sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

print("Chosen mean:", mean)
print("Sample mean:", sample_mean)

print("\nChosen standard deviation:", std)
print("Sample standard deviation:", sample_std)

#Histogram
plt.hist(data, bins=30, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normal Distribution")

plt.show()
```

## Task 5 - Filtering and Querying
Use Boolean indexing and/or query() to answer:
● Male passengers older than 50: how many?
● Female first-class passengers: how many, and what percentage survived?
● Age 20-40, Fare above overall median, and survived: how many?
● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(male_over_50))

female_first = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]
count = len(female_first)
survival_percentage = female_first["Survived"].mean() * 100
print("Female first-class passengers:", count)
print("Survival percentage:", survival_percentage)

fare_median = df["Fare"].median()
result = df[
    (df["Age"].between(20, 40)) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]
print("Count:", len(result))

result = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]
print("Count:", len(result))

southampton = df[df["Embarked"] == "S"]
southampton_median = southampton["Fare"].median()
result = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]
print("Southampton median fare:", southampton_median)
print("Count:", len(result))
```

## Task 6 - groupby() and Aggregation
● Survival rate by Sex.
● Survival rate by Pclass.
● Average Age and Fare by Pclass.
● Passenger count and survival rate by Sex-Pclass.
● Passenger count, average Fare, and survival rate by Embarked.

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean()
print(survival_by_sex)
print(survival_by_sex * 100)
print(df.groupby("Pclass")["Survived"].mean() * 100)
print(
    df.groupby("Pclass")[["Age", "Fare"]].mean()
)
result = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)
result["Survival_Rate"] *= 100
print(result)
result = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

result["Survival_Rate"] *= 100

print(result)
```

## Task 7 - Missing Values and Fare Outliers
● Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
● Fill missing Age values with the mean Age. Show missing counts before and after.
● Try to fill NaN values by imputing using: mean, median, mode, random value
● For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python
missing_count = df.isnull().sum()
print(missing_count)

missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)

#Create table
missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})
print(missing_table)

df.isnull().sum().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Count")

plt.show()

print("Before:")
print(df["Age"].isnull().sum())
age_mean = df["Age"].mean()

df["Age"] = df["Age"].fillna(age_mean)
print("After:")
print(df["Age"].isnull().sum())
#Mean
df["Age_mean"] = df["Age"].fillna(df["Age"].mean())
#Median
df["Age_median"] = df["Age"].fillna(df["Age"].median())
#Mode
df["Age_mode"] = df["Age"].fillna(df["Age"].mode()[0])
#Random Value
age_random = df["Age"].dropna().sample(
    n=df["Age"].isna().sum(),
    random_state=42
).values

df_random = df["Age"].copy()

df_random.loc[df_random.isna()] = age_random

print(df_random.isnull().sum())
#Outliers
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Number of Fare outliers:", len(outliers))
```

## Task 8 - Features and Pivot Table
● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
print(df[["SibSp", "Parch", "FamilySize"]].head())
df["IsAlone"] = np.where(df["FamilySize"] == 1, 1, 0)

print(df[["FamilySize", "IsAlone"]].head())
pivot = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print(pivot)
print(pivot * 100)
print("Highest survival group:")
print(pivot.stack().idxmax())
print("Survival rate:", pivot.stack().max() * 100)
print("\nLowest survival group:")
print(pivot.stack().idxmin())
print("Survival rate:", pivot.stack().min() * 100)
```
